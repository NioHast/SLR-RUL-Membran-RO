```python
# Cell 2: re-run marker on the 3 mismatched PDFs

import os, glob, subprocess, shutil
# Output directory on Kaggle
OUT="/kaggle/working/out"; os.makedirs(OUT, exist_ok=True)
# Find every PDF provided by the dataset
pdfs=sorted(glob.glob("/kaggle/input/**/*.pdf", recursive=True))
print("PDFs:",len(pdfs))

# Convert each PDF and print its return code
for f in pdfs:
    base=os.path.splitext(os.path.basename(f))[0]
    d=os.path.join(OUT,base); os.makedirs(d, exist_ok=True)
    r=subprocess.run(["marker_single", f, "--output_dir", d], capture_output=True, text=True)
    print(base,"rc",r.returncode, flush=True)

# Zip the outputs for download
shutil.make_archive("/kaggle/working/fix3_out","zip",OUT)
print(os.listdir("/kaggle/working"))
```
